```python
import os

def convert_file(filepath):
    with open(filepath, 'r') as f:
        lines = f.readlines()

    if not lines:
        return False

    new_lines = []
    changed = False

    for line in lines:
        parts = line.strip().split()
        if not parts:
            continue

        cls = parts[0]
        coords = list(map(float, parts[1:]))

        if len(coords) == 4:
            new_lines.append(line.strip())
            continue

        if len(coords) < 6 or len(coords) % 2 != 0:
            print(f"  WARNING: {filepath} has an unexpected number of values ({len(coords)}), skipping this line")
            new_lines.append(line.strip())
            continue

        xs = coords[0::2]
        ys = coords[1::2]

        x_min, x_max = min(xs), max(xs)
        y_min, y_max = min(ys), max(ys)
        x_min, x_max = max(0.0, x_min), min(1.0, x_max)
        y_min, y_max = max(0.0, y_min), min(1.0, y_max)

        x_center = (x_min + x_max) / 2
        y_center = (y_min + y_max) / 2
        width = x_max - x_min
        height = y_max - y_min

        new_lines.append(f"{cls} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}")
        changed = True

    with open(filepath, 'w') as f:
        f.write("\n".join(new_lines) + ("\n" if new_lines else ""))

    return changed


def convert_folder(label_dir):
    txt_files = [f for f in os.listdir(label_dir) if f.endswith('.txt')]
    converted_count = 0
    for fname in txt_files:
        path = os.path.join(label_dir, fname)
        if convert_file(path):
            converted_count += 1
    print(f"Done. Converted {converted_count} out of {len(txt_files)} label files in {label_dir}")
```

```python
convert_folder('/content/Aquafina.v1-first.yolov8/train/labels')
convert_folder('/content/Aquafina.v1-first.yolov8/valid/labels')
convert_folder('/content/Aquafina.v1-first.yolov8/test/labels')
```

```python
import os

for split in ['train', 'valid', 'test']:
    label_dir = f'/content/Aquafina.v1-first.yolov8/{split}/labels'
    bad_files = []

    for fname in os.listdir(label_dir):
        path = os.path.join(label_dir, fname)
        with open(path) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) not in (0, 5):
                    bad_files.append(fname)

    print(f"{split}: {len(bad_files)} remaining problem files")
```
